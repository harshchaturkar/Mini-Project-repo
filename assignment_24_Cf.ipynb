```python
x = df.drop("deposit", axis=1)
y = df["deposit"]

print("Independant feature :")
print(x.columns)

print("\nDependant feature :")
print(y.head)
```

```python
cat_cols = df.select_dtypes(include="object").columns
num_cols = df.select_dtypes(exclude="object").columns

print("Categorical columns :")
print(cat_cols)

print("\nNumerical columns :")
print(num_cols)
#cat_cols = cat_cols.drop("deposit")
```

```python
encoded = pd.get_dummies(df[cat_cols])
encoded = encoded.astype(int)
encoded
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
scaled = scaler.fit_transform(df[num_cols])
scaled = pd.DataFrame(scaled, columns=num_cols)
scaled
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(encoded, y, test_size=0.2, random_state=42)
```

**Logistic Regression**

```python
from sklearn.linear_model import LogisticRegression

model_LOR = LogisticRegression()
model_LOR.fit(x_train, y_train)
```

```python
pred_LOR = model_LOR.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
print("Accuracy Score :")
print(accuracy_score(y_test, pred_LOR))
print("\nConfusion Matrix :")
print(confusion_matrix(y_test, pred_LOR))
print("\nClassification Report :")
print(classification_report(y_test, pred_LOR))
```

**Decision Tree Classifier**

```python
from sklearn.tree import DecisionTreeClassifier

model_DT = DecisionTreeClassifier()
model_DT.fit(x_train, y_train)
```

```python
pred_DT = model_DT.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
print("Accuracy Score :")
print(accuracy_score(y_test, pred_DT))
print("\nConfusion Matrix :")
print(confusion_matrix(y_test, pred_DT))
print("\nClassification Report :")
print(classification_report(y_test, pred_DT))
```

**Support Vector Machine (SVM)**

```python
from sklearn.svm import SVC

model_SVC = SVC()
model_SVC.fit(x_train, y_train)
```

```python
pred_SVM = model_SVC.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
print("Accuracy Score :")
print(accuracy_score(y_test, pred_SVM))
print("\nConfusion Matrix :")
print(confusion_matrix(y_test, pred_SVM))
print("\nClassification Report :")
print(classification_report(y_test, pred_SVM))
```

**K-Nearest Neighbors (KNN)**

```python
from sklearn.neighbors import KNeighborsClassifier

model_KNN = KNeighborsClassifier(n_neighbors=5)
model_KNN.fit(x_train, y_train)
```

```python
pred_KNN = model_KNN.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
print("Accuracy Score :")
print(accuracy_score(y_test, pred_KNN))
print("\nConfusion Matrix :")
print(confusion_matrix(y_test, pred_KNN))
print("\nClassification Report :")
print(classification_report(y_test, pred_KNN))
```

**Naive Bayes**

```python
from sklearn.naive_bayes import GaussianNB

model_GNB = GaussianNB()
model_GNB.fit(x_train, y_train)
```

```python
pred_NB = model_GNB.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
print("Accuracy Score :")
print(accuracy_score(y_test, pred_NB))
print("\nConfusion Matrix :")
print(confusion_matrix(y_test, pred_NB))
print("\nClassification Report :")
print(classification_report(y_test, pred_NB))
```

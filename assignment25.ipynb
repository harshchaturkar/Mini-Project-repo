```python
import pandas as pd
from sklearn.model_selection import GridSearchCV
from sklearn.svm import SVC

model = SVC()
param_grid = {
    'C': [1, 10, 20],
    'kernel': ['linear', 'rbf']
}

grid = GridSearchCV(
    estimator=model,
    param_grid=param_grid,
    cv=5
)

grid.fit(X_train, y_train)
results = pd.DataFrame(grid.cv_results_)

print(results[['param_C', 'param_kernel', 'mean_test_score']])
print("\nBest Parameters:", grid.best_params_)
print("Best Score:", grid.best_score_)
```

```python
from sklearn.model_selection import RandomizedSearchCV
from sklearn.svm import SVC

model = SVC()
param_grid = {
    'C': [1, 10, 20],
    'kernel': ['linear', 'rbf']
}

random = RandomizedSearchCV(
    estimator=model,
    param_distributions=param_grid,
    n_iter=5,
    cv=5,
    random_state=42
)

random.fit(X_train, y_train)
results = pd.DataFrame(random.cv_results_)

print(results[['param_C', 'param_kernel', 'mean_test_score']])

print("\nBest Parameters:", random.best_params_)
print("Best Score:", random.best_score_)

print("\nComparison:")
print("Randomized Search checks only 5 random combinations, so it is faster.")
print("Grid Search checks all combinations, so it is slower but may find the best result.")
```

```python
import pandas as pd
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier, AdaBoostClassifier, GradientBoostingClassifier
from xgboost import XGBClassifier

iris = load_iris()
X = iris.data
y = iris.target

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,
    stratify=y,
    random_state=42
)

models = {
    "SVM": SVC(C=10, kernel="linear"),
    "Random Forest": RandomForestClassifier(n_estimators=100, random_state=42),
    "AdaBoost": AdaBoostClassifier(n_estimators=100, random_state=42),
    "Gradient Boosting": GradientBoostingClassifier(
        n_estimators=100,
        learning_rate=0.1,
        random_state=42
    ),
    "XGBoost": XGBClassifier(
        n_estimators=100,
        learning_rate=0.1,
        max_depth=3,
        random_state=42
    )
}

results = []

for name, model in models.items():
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    accuracy = accuracy_score(y_test, y_pred)

    results.append([name, accuracy])

comparison = pd.DataFrame(results, columns=["Model", "Accuracy"])

print(comparison)

best = comparison.loc[comparison["Accuracy"].idxmax()]

print("\nBest Model:", best["Model"])
print("Best Accuracy:", best["Accuracy"])
```

```python
import pandas as pd
import joblib

from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.model_selection import GridSearchCV
from sklearn.model_selection import RandomizedSearchCV

from sklearn.svm import SVC
from sklearn.ensemble import (
    RandomForestClassifier,
    AdaBoostClassifier,
    GradientBoostingClassifier
)

from sklearn.metrics import accuracy_score
from xgboost import XGBClassifier

iris = load_iris()
X = iris.data
y = iris.target

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)


best_manual_model = None
best_manual_accuracy = 0

for C in [1, 10, 20]:
    for kernel in ["linear", "rbf"]:

        model = SVC(C=C, kernel=kernel)
        model.fit(X_train, y_train)
        pred = model.predict(X_test)
        acc = accuracy_score(y_test, pred)

        print("Manual Search")
        print("C =", C)
        print("Kernel =", kernel)
        print("Accuracy =", acc)
        print()

        if acc > best_manual_accuracy:
            best_manual_accuracy = acc
            best_manual_model = model


param_grid = {
    "C": [1, 10, 20],
    "kernel": ["linear", "rbf"]
}

grid = GridSearchCV(
    SVC(),
    param_grid,
    cv=5
)

grid.fit(X_train, y_train)
grid_model = grid.best_estimator_
grid_pred = grid_model.predict(X_test)
grid_accuracy = accuracy_score(y_test, grid_pred)

print("Grid Best Parameters:", grid.best_params_)
print("Grid Accuracy:", grid_accuracy)


random = RandomizedSearchCV(
    SVC(),
    param_distributions=param_grid,
    n_iter=5,
    cv=5,
    random_state=42
)

random.fit(X_train, y_train)
random_model = random.best_estimator_
random_pred = random_model.predict(X_test)
random_accuracy = accuracy_score(y_test, random_pred)

print("Random Best Parameters:", random.best_params_)
print("Random Accuracy:", random_accuracy)


rf = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

rf.fit(X_train, y_train)
rf_pred = rf.predict(X_test)
rf_accuracy = accuracy_score(y_test, rf_pred)


ada = AdaBoostClassifier(
    n_estimators=100,
    random_state=42
)

ada.fit(X_train, y_train)
ada_pred = ada.predict(X_test)
ada_accuracy = accuracy_score(y_test, ada_pred)


gb = GradientBoostingClassifier(
    n_estimators=100,
    learning_rate=0.1,
    random_state=42
)

gb.fit(X_train, y_train)
gb_pred = gb.predict(X_test)
gb_accuracy = accuracy_score(y_test, gb_pred)


xgb = XGBClassifier(
    n_estimators=100,
    learning_rate=0.1,
    max_depth=3,
    random_state=42
)

xgb.fit(X_train, y_train)
xgb_pred = xgb.predict(X_test)
xgb_accuracy = accuracy_score(y_test, xgb_pred)


comparison = pd.DataFrame({

    "Model": [

        "Manual Search SVM",
        "Grid Search SVM",
        "Random Search SVM",
        "Random Forest",
        "AdaBoost",
        "Gradient Boosting",
        "XGBoost"

    ],

    "Accuracy": [

        best_manual_accuracy,
        grid_accuracy,
        random_accuracy,
        rf_accuracy,
        ada_accuracy,
        gb_accuracy,
        xgb_accuracy

    ]

})

print("\nComparison Table\n")
print(comparison)

models = {

    "Manual Search SVM": best_manual_model,
    "Grid Search SVM": grid_model,
    "Random Search SVM": random_model,
    "Random Forest": rf,
    "AdaBoost": ada,
    "Gradient Boosting": gb,
    "XGBoost": xgb

}

best_model_name = comparison.loc[
    comparison["Accuracy"].idxmax(),
    "Model"
]

best_model = models[best_model_name]
print("\nBest Model:", best_model_name)
joblib.dump(best_model, "best_model.pkl")
print("\nModel Saved Successfully")
```

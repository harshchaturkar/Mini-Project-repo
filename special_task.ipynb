```python
import numpy as np
import pandas as pd
import joblib
```

```python
df = pd.read_csv("./student_performance_dataset.csv")
df
```

```python
df.describe()
```

```python
x = df.drop(["final_grade","final_exam_score"], axis=1)
y = df["final_grade"]
print("Independent features :")
print(x)
print("Dependent features :")
print(y)
```

```python
cat_cols = df.select_dtypes(include=["object"]).columns
num_cols = df.select_dtypes(include=["int64","float64"]).columns
print("Categorical colums")
print(cat_cols)
print("Numerical columns")
print(num_cols)
#cat_cols = df.drop("final_grade",axis=1)
#num_cols = df.drop("student_id",axis=1)
```

```python
encoded = pd.get_dummies(df[cat_cols])
encoded = encoded.astype(int)
encoded
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
scaled = scaler.fit_transform(df[num_cols])
scaled = pd.DataFrame(scaled,columns=num_cols)
scaled
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(encoded, y, test_size=0.20, random_state=42)
```

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression()
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
print("Predected value:")
print(y_pred[:10])
print("\nActual value:")
print(y_test[:10])
```

```python
joblib.dump(model,"student_performance_dataset.pkl")
joblib.dump(scaler,"scaler.pkl")
joblib.dump(encoded.columns.tolist(), "columns.pkl")
print("Model and preprocessing objects saved successfully!")
```

```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv("./heart.csv")
df
```

```python
x = df.drop("HeartDisease", axis=1)
y = df["HeartDisease"]
print("Independent features :")
print(x)
print("\nDependent features :")
print(y)
```

```python
encoded = pd.get_dummies(x)
encoded = encoded.astype(int)
encoded
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(encoded, y, test_size=0.20,  random_state=42)

print("Shape of X_train :", X_train.shape)
print("Shape of X_test :", X_test.shape)
print("Shape of y_train :", X_train.shape)
print("Shape of y_test :", X_test.shape)
```

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression()
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
print("First 10 actual values :")
print(y_test.head(10))
print("\nFirst 10 predicted values :")
print(y_pred[:10])
```

```python
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(y_pred, y_test)
print(cm)

print("\nTrue Negative :",cm[0][0])
print("\nFalse Positive :",cm[0][1])
print("\nFalse Negative :",cm[1][0])
print("\nTrue Positive :",cm[1][1])
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, classification_report

print("Accuracy :", accuracy_score(y_pred, y_test))
print("\nPrecision :", precision_score(y_pred, y_test))
print("\nRecall :", recall_score(y_pred, y_test))
print("\nF1 score :", f1_score(y_pred, y_test))

print("\nClassification Report:")
print(classification_report(y_test, y_pred))
```

```python
import joblib

joblib.dump(model, "heart.pkl")
joblib.dump(encoded.columns.tolist(), "columns.pkl")

print("Model and columns saved successfully!")
```

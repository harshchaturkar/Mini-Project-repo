```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('./car_price_dataset.csv')
df
```

```python
x = df.drop('Price',axis=1)
y = df['Price']

print(" Independent Features (X) :")
print(x.columns)
print("\nDependent Feature (Y) :")
print(y.name)
print("\nShape of x :")
print(x.shape)
print("Shape of y :")
print(y.shape)
```

```python
cat_cols = df.select_dtypes('int64','float64').columns
print(" categorical columns ")
cat_cols
```

```python
encoded = pd.get_dummies(df[cat_cols])
encoded
convert =encoded.astype(int)
convert
convert.head()
```

```python
from sklearn.preprocessing import StandardScaler

numeric_cols = ['Model_Year', 'Mileage', 'Doors', 'Owner_Count', 'Horsepower']

scaler = StandardScaler()

scaled_features = scaler.fit_transform(df[numeric_cols])

scaled_df = pd.DataFrame(scaled_features, columns=numeric_cols)

print(scaled_df.head())
```

```python
from sklearn.model_selection import train_test_split

# Train-Test Split
X_train, X_test, y_train, y_test = train_test_split(
    x, y, test_size=0.33, random_state=42
)

# Print shapes
print("X_train shape:", X_train.shape)
print("X_test shape :", X_test.shape)
print("y_train shape:", y_train.shape)
print("y_test shape :", y_test.shape)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

X_train, X_test, y_train, y_test = train_test_split(convert, y, test_size=0.33, random_state=42)

model = LinearRegression()
model.fit(X_train, y_train)
# Print intercept
print("Intercept:", model.intercept_)

# Print coefficients
print("\nCoefficients:", model.coef_)
```

```python
y_pred = model.predict(X_test)

print("First 10 Predicted Values:")
print(y_pred[:10])

print("\nFirst 10 Actual Values:")
print(y_test.iloc[:10])

comparison = pd.DataFrame({
    "Actual": y_test.iloc[:10].values,
    "Predicted": y_pred[:10]
})

print("\nComparison of Actual vs Predicted:")
print(comparison)
```

```python
from sklearn.metrics import r2_score

r2 = r2_score(y_test, y_pred)

print("R² Score:", r2)
```

```python
#The model explains 56.74% of the variation in car prices.
#The model performance is moderate (average).
#The predictions are reasonably accurate, but there is room for improvement.
```

```python
import joblib

# Save the trained Linear Regression model
joblib.dump(model, "LR_ford_car.pkl")

# Save the StandardScaler object
joblib.dump(scaler, "scaler.pkl")

# Save the list of feature columns
joblib.dump(convert.columns.tolist(), "columns.pkl")

print("Model and preprocessing objects saved successfully!")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
plt.subplot(2,2,1)
sns.histplot(data['age'])
plt.title("Age distribution")
plt.subplot(2,2,2)
sns.histplot(data['bmi'])
plt.title("BMI distribution")
plt.subplot(2,2,3)
sns.histplot(data['children'])
plt.title("Children distribution")
plt.subplot(2,2,4)
sns.histplot(data['expenses'])
plt.title("Expenses distribution")

plt.tight_layout()
```

```python
plt.subplot(2,2,1)
sns.countplot(data['sex'])
plt.title("Sex count")
plt.subplot(2,2,2)
sns.countplot(data['smoker'])
plt.title("Smoker count")
plt.subplot(2,2,3)
sns.histplot(data['region'])
plt.title("region count")

plt.tight_layout()
```

```python
data_riletion = data[['age','bmi','children','expenses']].corr()
sns.heatmap(data_riletion,annot=True)
plt.title("Correlation Heatmap of numerical variables")
```

```python
avg_charges = np.mean(data['expenses'])
avg_charges
print("Average of insurance charges :",avg_charges)

max_charges = np.max(data['expenses'])
min_charges = np.min(data['expenses'])
print("Maximum Charges :",max_charges)
print("Minimum Charges :",min_charges)


avg_expenses = data.groupby('smoker')['expenses'].mean()
print(avg_expenses)
```

```python
sns.boxplot(x=data['smoker'], y=data['expenses'], hue=data['sex'])
```

```python
avg_age = np.mean(data['age'])
avg_bmi = np.mean(data['bmi'])
print('Average of Age :',avg_age)
print('Average of BMI :',avg_bmi)

avg_expenses = data.groupby('smoker')['expenses'].mean()
print('\n',avg_expenses)

print('\nNumber of customers in regions')
print(data['region'].value_counts())

print("\nVisualization Summary:")
print("- Southeast has the highest number of customers.")
print("- Most customers belong to the Southeast region.")
print("- The Northeast, Northwest, and Southwest regions have similar customer counts.")
```

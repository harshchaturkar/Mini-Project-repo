```python
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test =train_test_split(encoded, y, test_size=0.20, random_state=42)

model_LOR = LogisticRegression()
model_LOR.fit(X_train, y_train)

model_KNN = KNeighborsClassifier(n_neighbors=5)
model_KNN.fit(X_train, y_train)

model_NB = GaussianNB()
model_NB.fit(X_train, y_train)

pred_LOR = model_LOR.predict(X_test)
pred_KNN = model_KNN.predict(X_test)
pred_NB = model_NB.predict(X_test)


# Create comparison table
results = pd.DataFrame({
    "Algorithm": [
        "Logistic Regression",
        "KNN (k=5)",
        "Naive Bayes"
    ],
    "Accuracy": [
        accuracy_score(y_test, pred_LOR),
        accuracy_score(y_test, pred_KNN),
        accuracy_score(y_test, pred_NB)
    ],
    "Precision": [
        precision_score(y_test, pred_LOR),
        precision_score(y_test, pred_KNN),
        precision_score(y_test, pred_NB)
    ],
    "Recall": [
        recall_score(y_test, pred_LOR),
        recall_score(y_test, pred_KNN),
        recall_score(y_test, pred_NB)
    ],
    "F1 Score": [
        f1_score(y_test, pred_LOR),
        f1_score(y_test, pred_KNN),
        f1_score(y_test, pred_NB)
    ]
})
print("Comparison Table:")
print("\n",results)

best_model = results.loc[results["Accuracy"].idxmax()]

print("\nBest Model:")
print(best_model)
```

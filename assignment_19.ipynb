```python

# Numeric columns
cols = ["num_rear_camera", "num_front_camera", "ratings", "num_of_ratings", "sales"]

# Create subplots
plt.figure(figsize=(12,8))

for i, col in enumerate(cols, 1):
    plt.subplot(2, 3, i)
    plt.hist(df[col], bins=10, edgecolor="black")
    plt.title(col)
    plt.xlabel(col)
    plt.ylabel("Count")

plt.tight_layout()
plt.show()
```

```python
#num_rear_camera: Most phones have 3–4 rear cameras.
#num_front_camera: Most phones have 1 front camera.
#ratings: Most phones have high ratings (around 4.3–4.4).
#num_of_ratings: A few phones have many ratings, while most have fewer.
#sales: Most phones have low sales, and only a few have very high sales
```

```python

# Count plots
sns.countplot(x="brand", data=df)
plt.show()

sns.countplot(x="model", data=df)
plt.xticks(rotation=90)
plt.show()

sns.countplot(x="base_color", data=df)
plt.show()
```

```python
# Correlation heatmap
plt.figure(figsize=(6,5))
sns.heatmap(df.corr(numeric_only=True), annot=True)
plt.title("Correlation Heatmap")
plt.show()
```
